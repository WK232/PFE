```python
import matplotlib
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator
from mpl_toolkits.axes_grid1.inset_locator import mark_inset
from mpl_toolkits.axes_grid1.inset_locator import inset_axes

# 设置Matplotlib样式
matplotlib.rcParams['font.family'] = 'serif'
matplotlib.rcParams['font.serif'] = ['Times New Roman'] + matplotlib.rcParams['font.serif']
matplotlib.rcParams['font.size'] = 10
matplotlib.rcParams['text.usetex'] = True

# 数据定义
NAS0_0 = [39.77] * 10 + [75.21] * 22 + [39.77] * 18
NAS0_1 = [39.77] + [75.21] + [39.77] * 3 + [75.21] * 34 + [39.77] * 11
NAS0_2 = [39.77] * 10 + [75.21] * 5 + [90.51] + [87.00] * 20 + [75.21] * 3 + [39.77] * 11
NAS0_3 = [39.77] * 2 + [75.21] * 46 + [39.77] * 2
NAS0 = [NAS0_0, NAS0_1, NAS0_2, NAS0_3]
NAS0 = np.array(NAS0).T

NAS2_0 = [39.77] * 2 + [75.21] + [90.51] * 3 + [91.21] * 2 + [91.55] * 42
NAS2_1 = [39.77] * 4 + [75.21] + [90.86] + [91.12] * 2 + [91.55] * 42
NAS2_2 = [85.13] + [73.55] + [75.21] * 2 + [90.51] * 7 + [91.12] + [91.55] * 38
NAS2_3 = [39.77] * 5 + [87.00] + [87.30] + [91.55] * 43
NAS2 = [NAS2_0, NAS2_1, NAS2_2, NAS2_3]
NAS2 = np.array(NAS2).T

NASp1_1 = [86.55] + [90.24] * 2 + [89.98] * 2 + [90.2] * 3 + [90.57] * 7 + [91.55] + [90.57] + [91.55] + [90.57] * 4 + [91.55] * 13 + [90.57] * 12 + [90.40] * 3
NASp1_2 = [90.19] * 2 + [90.20] + [90.57] * 4 + [90.44] * 4 + [90.57] * 3 + [90.44] * 4 + [90.57] * 10 + [91.55] * 12 + [90.57] * 10
NASp1_3 = [89.56] + [90.32] * 5 + [90.12] + [90.32] * 4 + [91.61] + [91.55] * 11 + [91.12] * 16 + [91.5] * 11
NASp1_4 = [86.54] + [89.4] + [89.86] + [90.2] * 3 + [90.57] * 15 + [91.55] * 26 + [91.12] * 3
NASp1 = [NASp1_1, NASp1_2, NASp1_3, NASp1_4]
NASp1 = np.array(NASp1).T

NASp2_1 = [89.65] + [88.96] + [89.61] + [90.2] * 2 + [89.98] * 2 + [90.2] * 20 + [90.57] * 23
NASp2_2 = [89.15] + [90.18] + [90.57] + [90.2] * 6 + [90.57] * 41
NASp2_3 = [90.06] + [89.97] + [86.66] + [89.21] + [90.44] + [90.57] * 4 + [90.4] * 10 + [90.57] * 23 + [90.4] * 8
NASp2_4 = [89.77] + [90.2] * 4 + [89.98] * 3 + [90.2] * 8 + [90.57] * 21 + [90.32] * 13
NASp2 = [NASp2_1, NASp2_2, NASp2_3, NASp2_4]
NASp2 = np.array(NASp2).T

means2 = np.mean(NAS2, axis=1)
stds2 = np.std(NAS2, axis=1, ddof=1)
means0 = np.mean(NAS0, axis=1)
stds0 = np.std(NAS0, axis=1, ddof=1)
meansp1 = np.mean(NASp1, axis=1)
stdsp1 = np.std(NASp1, axis=1, ddof=1)
meansp2 = np.mean(NASp2, axis=1)
stdsp2 = np.std(NASp2, axis=1, ddof=1)

half_widths0 = 1.96 * stds0 / 2  # 95%置信区间的半宽度
lower_bounds0 = means0 - half_widths0
upper_bounds0 = means0 + half_widths0
half_widths2 = 1.96 * stds2 / 2  # 95%置信区间的半宽度
lower_bounds2 = means2 - half_widths2
upper_bounds2 = means2 + half_widths2

half_widthsp1 = 1.96 * stdsp1 / 2  # 95%置信区间的半宽度
lower_boundsp1 = meansp1 - half_widthsp1
upper_boundsp1 = meansp1 + half_widthsp1

half_widthsp2 = 1.96 * stdsp2 / 2  # 95%置信区间的半宽度
lower_boundsp2 = meansp2 - half_widthsp2
upper_boundsp2 = meansp2 + half_widthsp2

# Early stage data
NASpe_1 = [89.43, 90.11, 89.61] + [90.20] * 8 + [90.57]*14 + [91.55] * 12 + [91.12] * 13
NASpe_2 = [89.90] + [90.20] * 16 + [90.57]*20 + [90.40] * 6 + [90.12] * 7
NASpe_3 = [89.33, 89.60, 90.57] + [90.20] * 14 + [90.57] * 23 + [90.44] * 2 + [90.10] + [90.14] * 5 + [89.81] * 2
NASpe_4 = [90.27, 90.11] + [90.20] * 17 +[90.57]*3 + [91.55] * 11 + [91.12] * 7 + [90.51]*10
early = [NASpe_1, NASpe_2, NASpe_3, NASpe_4]
early = np.array(early).T
means_early = np.mean(early, axis=1)
stds_early = np.std(early, axis=1, ddof=1)
half_widths_early  = 1.96 * stds_early  / 2  # 95%置信区间的半宽度
lower_bounds_early  = means_early  - half_widths_early 
upper_bounds_early  = means_early  + half_widths_early 

# Middle stage data
NASpm_1 = [39.77]*3 + [75.21] * 14 + [90.51, 91.12] + [91.55] * 21 + [91.12] * 10
NASpm_2 = [73.55] * 5 + [75.21] * 12 + [91.12] * 5 + [91.55]*15 + [91.12] * 13
NASpm_3 = [39.77] * 3 + [75.21] * 12 + [85.07] * 2 + [91.12] * 2 + [91.55] * 24 + [91.12] * 6 + [90.51]
NASpm_4 = [73.55] + [87.00] * 2 + [75.21] * 14 + [90.51] * 2 + [91.21] + [91.55] * 23 + [91.50] * 7
middle = [NASpm_1, NASpm_2, NASpm_3, NASpm_4]
middle = np.array(middle).T
means_middle = np.mean(middle, axis=1)
stds_middle = np.std(middle, axis=1, ddof=1)
half_widths_middle  = 1.96 * stds_middle  / 2  # 95%置信区间的半宽度
lower_bounds_middle  = means_middle  - half_widths_middle 
upper_bounds_middle  = means_middle  + half_widths_middle 

# Late stage data
NASpl_1 = [39.77] * 8 + [75.21] * 28 + [87.00] * 2 + [90.86] * 2 + [91.12] * 10
NASpl_2 = [73.55] + [39.77] * 8 + [75.21] * 26 + [85.07] * 2 + [90.51] + [91.12] * 3 + [91.55] * 9
NASpl_3 = [39.77] * 12 + [75.21] * 23 + [85.07, 90.51] + [91.12] * 3 + [91.55] * 10
NASpl_4 = [39.77] * 5 + [75.21] * 26 + [39.77] * 3 + [75.21] + [90.51] * 6 + [91.12] * 4 + [91.55] * 4 + [91.50]
late = [NASpl_1, NASpl_2, NASpl_3, NASpl_4]
late = np.array(late).T
means_late = np.mean(late, axis=1)
stds_late = np.std(late, axis=1, ddof=1)
half_widths_late  = 1.96 * stds_late  / 2  # 95%置信区间的半宽度
lower_bounds_late  = means_late  - half_widths_late 
upper_bounds_late  = means_late  + half_widths_late 

# 创建图表和轴
fig, ax = plt.subplots(figsize=(9, 6))

# 绘制主图
ax.plot(means0, label='$p = 0$', color='blue')
ax.fill_between(np.arange(len(means0)), lower_bounds0, upper_bounds0, color='blue', alpha=0.1)
ax.plot(meansp1, label='$p = 0.1$', color='green')
ax.fill_between(np.arange(len(meansp1)), lower_boundsp1, upper_boundsp1, color='green', alpha=0.1)
ax.plot(meansp2, label='$p = 0.2$', color='orange')
ax.fill_between(np.arange(len(meansp2)), lower_boundsp2, upper_boundsp2, color='orange', alpha=0.1)
ax.plot(means2, label='$p_{\max} = 0.2$', color='red')
ax.fill_between(np.arange(len(means2)), lower_bounds2, upper_bounds2, color='red', alpha=0.1)
ax.plot(means_early, label='$early$', color='lightseagreen')  # 使用 lightseagreen 颜色
ax.fill_between(np.arange(len(means_early)), lower_bounds_early, upper_bounds_early, color='lightseagreen', alpha=0.1)
ax.plot(means_middle, label='$middle$', color='magenta')
ax.fill_between(np.arange(len(means_middle)), lower_bounds_middle, upper_bounds_middle, color='magenta', alpha=0.1)
ax.plot(means_late, label='$late$', color='purple')
ax.fill_between(np.arange(len(means_late)), lower_bounds_late, upper_bounds_late, color='purple', alpha=0.1)

# 设置X轴刻度
ax.xaxis.set_ticks_position("bottom")
ax.xaxis.set_label_position("bottom")
ax.xaxis.set_major_locator(MultipleLocator(base=10))
ax.set_xlabel('Epoch')
ax.set_ylabel('CIFAR-10 Valid Accuracy (\%)')
ax.legend(loc='best')

# 插入放大图
axins = ax.inset_axes((0.65, 0.6, 0.35, 0.23))
axins.plot(meansp1, label='$p = 0.1$', color='green')
axins.fill_between(np.arange(len(meansp1)), lower_boundsp1, upper_boundsp1, color='green', alpha=0.1)
axins.plot(meansp2, label='$p = 0.2$', color='orange')
axins.fill_between(np.arange(len(meansp2)), lower_boundsp2, upper_boundsp2, color='orange', alpha=0.1)
axins.plot(means2, label='$p_{\max} = 0.2$', color='red')
axins.fill_between(np.arange(len(means2)), lower_bounds2, upper_bounds2, color='red', alpha=0.1)
axins.plot(means_early, label='$early$', color='lightseagreen')  # 使用 lightseagreen 颜色
axins.fill_between(np.arange(len(means_early)), lower_bounds_early, upper_bounds_early, color='lightseagreen', alpha=0.1)
axins.plot(means_middle, label='$middle$', color='magenta')
axins.fill_between(np.arange(len(means_middle)), lower_bounds_middle, upper_bounds_middle, color='magenta', alpha=0.1)
axins.plot(means_late, label='$late$', color='purple')
axins.fill_between(np.arange(len(means_late)), lower_bounds_late, upper_bounds_late, color='purple', alpha=0.1)
axins.set_xlim(40, 50)
axins.set_ylim(90.3, 91.6)
mark_inset(ax, axins, loc1=1, loc2=2, fc="none", ec='k', linestyle='dashed', lw=1)

# 调整布局
plt.tight_layout()

# 保存图片
plt.savefig('accuracy_NAS.pdf', format='pdf', dpi=600)

# 显示图表
plt.show()
```

```python
import sys
# |nor_conv_3x3~0|+|nor_conv_3x3~0|nor_conv_3x3~1|+|skip_connect~0|nor_conv_3x3~1|nor_conv_3x3~2|
from graphviz import Digraph


g = Digraph(
    format='pdf',
    edge_attr=dict(fontsize='10', fontname="sans-serif"),
    node_attr=dict(style='filled', shape='rect', align='center', fontsize='10', height='0.5', width='0.5',
                   penwidth='2', fontname="sans-serif"),
    engine='dot')
g.body.extend(['rankdir=LR'])

g.node("0", fillcolor='darkseagreen2')
g.node("1", fillcolor='lightblue')
g.node("2", fillcolor='lightblue')
g.node("3", fillcolor='palegoldenrod')
g.edge("0", "1", label="nor_conv_3x3", fillcolor="gray")
g.edge("0", "2", label="nor_conv_3x3", fillcolor="gray")
g.edge("1", "2", label="nor_conv_3x3", fillcolor="gray")
g.edge("0", "3", label="skip_connect", fillcolor="gray")
g.edge("1", "3", label="nor_conv_3x3", fillcolor="gray")
g.edge("2", "3", label="nor_conv_3x3", fillcolor="gray")
filename = 'fig/genotypes/cnn/nas201_cifar10'


g.render(filename, view=True)
```
